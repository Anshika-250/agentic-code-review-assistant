```python
buggy_code = """
def sum_even(nums):
  total = 0
  for i in range(1,len(nums)):
    if nums[i]%2==0:
      total+=nums[i]
  return total
"""
test_cases = [
    {"input":[1,2,3,4,5,6],"expected":12},
    {"input":[2,4,6,8],"expected":20},
    {"input":[1,3,5],"expected":0},
    {"input":[],"expected":0}
]
```

```python
def run_tests(code_str,test_cases):
  namespace = {}
  exec(code_str,namespace)
  func = namespace["sum_even"]
  results = []
  for case in test_cases:
    try:
      actual = func(case["input"])
    except Exception as e:
        actual = f"ERROR:{e}"
    results.append({
          "input":case["input"],
          "expected": case["expected"],
          "actual": actual,
          "passed": actual == case["expected"]
      })
  return results

for r in run_tests(buggy_code, test_cases):
    print("PASS" if r["passed"] else "FAIL", r)
```
